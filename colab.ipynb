### 🚀 Step 1: launch the app

Run the cell below to start the **Gradio app** with a public share link.

```python
# CELL 2: launch the Gradio app with a share link
import sys
import threading

print("🚀 Starting folio...")

sys.path.insert(0, "hf")
from app import demo

def run():
    demo.launch(share=True, server_name="0.0.0.0", server_port=7860)

threading.Thread(target=run, daemon=True).start()

import time, urllib.request
time.sleep(3)
try:
    urllib.request.urlopen("http://localhost:7860")
except Exception:
    time.sleep(5)

print("✅ folio is running!")
print("🌐 Open the Gradio link below (the line 'Running on public URL:')")
```
